## PANDAS THEORY

The most basic data structures in Pandas are the `Series` and `DataFrame` objects. 

* Series

The series object holds a one-dimensional array, similar to a list. The values could
be of any type including python primitive datatypes and objects.

* DataFrame

The DataFrame object holds data like a two-dimensional array or a table with rows and columns.

### **Creating a Pandas Series**

We create an instance of a Pandas Series by calling the `Series` method of the `pandas` library,
and passing a list of values to it.

```python
s = pd.Series([4, 6, 8, np.nan, np.nan, 5])

s
```

### **Creating a Pandas DataFrame**

We would consider two methods of creating an instance of a Pandas DataFrame object.

1.   By passing a Numpy array with a datetime index using `date_range()` and labeled columns.

```python
dates = pd.date_range('20261030', periods=6)

dates
```

```python
df = pd.DataFrame(np.random.randn(6, 4), index=dates, columns=list('ABCD'))

df.describe()
```

2.  By passing a dictionary of objects where the `key` is taken as the label of the column.

```python
df2 = pd.DataFrame({
    'A': True,
    'B': pd.Timestamp('20261030'),
    'C': pd.Series(1, index=list(range(4)), dtype='float32'),
    'D': np.array([3] * 4, dtype='int32'),
    'E': pd.Categorical(['test', 'train', 'test', 'train']),
    'F': "foo"
})

df2
```

```python
data = {
    'first_name': ['John', 'Piper', 'Perry', 'Anais', 'Joseph'],
    'last_name': ['Bush', 'McKinney', 'Oswald', 'Woods', 'Smith'],
    'class': ['grade 8', 'grade 7', 'grade 7', 'grade 8', 'grade 8'],
    'test_scores': [20, 16, 14, 14, 18],
    'Passed': [True, True, False, False, True]
}

class_scores = pd.DataFrame(data)
class_scores['test_scores'].describe()
```
