## **Exercise 2: `Groupby()`**

```python
df_avg = df.groupby(by='ward', as_index=False).agg(
    mean_los=('los_days', 'mean'),
    mean_readmission=('readmitted', 'mean') 
)

df_avg
```

Comparing the values of this function with the values from the `summarise_by_ward()` method.

```python
def summarise_by_ward(records: list[dict]) -> dict:
    """ Return the mean length of stay and readmission rate in a dictionary.

        The mean LOS and readmission rate is matched to each ward using
        the ward name as the dict key. This function takes a `list` of `dict` 
        as the input and outputs a `dict` of the mean length of stay 
        and readmission rate for each ward.
        
        An empty list returns {} without an error, while a record missing 
        the `ward`, `los` or `readmitted` keys raises a `KeyError`
        for the first missing key it catches.   
        
        Booleans count as numbers.
    """
    if not records:
        return {}
    for record in records:
        if 'ward' not in record:
            raise KeyError('The ward key is missing from the dictionary')
        if 'los_days' not in record:
            raise KeyError('The los key is missing from the dictionary')
        if 'readmitted' not in record:
            raise KeyError('The readmitted key is missing from the dictionary')
    
    records_dict = {}
    overview = {}
    for record in records:
        ward = record['ward']
        los = record['los_days']
        readmitted = record['readmitted']

        if ward not in records_dict:
            records_dict[ward] = {'los': [], 'readmitted': []}
        records_dict[ward]['los'].append(los)
        records_dict[ward]['readmitted'].append(readmitted)

    for ward, data in records_dict.items():
        mean_los = np.mean(data['los'])
        readmission_rate = np.mean(data['readmitted'])
        overview[ward] = {'mean_los': mean_los, 'readmission_rate': readmission_rate}
        
    return overview
```

```python
# Converting the DataFrame to a dictionary object
records = df.to_dict("records")
records
```

**Assertions for `summarise_by_ward()`**

```python
result = summarise_by_ward(records)
assert math.isclose(result['A']['mean_los'], 4.6666666667), "The mean LOS for ward A is 4.6666666667"
assert math.isclose(result['B']['mean_los'], 8.0), "The mean LOS for ward B is 8.0"
assert math.isclose(result['C']['mean_los'], 8.5), "The mean LOS for ward C is 8.5"
assert math.isclose(result['A']['readmission_rate'], 2/3), "The readmission rate for ward A is 0.6666666667"
assert math.isclose(result['B']['readmission_rate'], 1/3), "The readmission rate for ward B is 0.3333333333"
assert math.isclose(result['C']['readmission_rate'], 1/2), "The readmission rate for ward C is 0.5"
```

**Assertions for Pandas `df.groupby()`**

```python
result_2 = df_avg.copy()
assert math.isclose(result_2['mean_los'][0], 4.6666666667), "The mean LOS for ward A is 4.6666666667"
assert math.isclose(result_2['mean_los'][1], 8.0), "The mean LOS for ward B is 8.0"
assert math.isclose(result_2['mean_los'][2], 8.5), "The mean LOS for ward C is 8.5"
assert math.isclose(result_2['mean_readmission'][0], 2/3), "The readmission rate for ward A is 0.6666666667"
assert math.isclose(result_2['mean_readmission'][1], 1/3), "The readmission rate for ward B is 0.3333333333"
assert math.isclose(result_2['mean_readmission'][2], 1/2), "The readmission rate for ward C is 0.5"
```

## **Exercise 3: Fill missing ages by ward**

```python
cleaned_df = df.copy()

cleaned_df['age'] = df['age'].fillna(df.groupby('ward')['age'].transform('median'))
cleaned_df
```

This exercise uses the `.fillna()` method to fill missing values in the 'age' column, using the groupby and transform methods to pass the median age of each ward (A, B, C) to the fillna method.

## **Exercise 4:** Age groups

```python
# bin edges and labels
df_grouped = cleaned_df.copy()
bins = [0, 40, 59, np.inf]
labels = ['under 40', '40-59', '60+']

df_grouped['age_group'] = pd.cut(df_grouped['age'], bins=bins, labels=labels, right=True)
df_grouped
```

The Above exercise uses the `pd.cut()` method to create a new column that categorizes the ages into 3 groups, using custom numerical boundaries (bins) and categorical values (labels). The right=True argument includes the right bin values or the edge intervals.

**Explanation for `.copy()`**

The reason the `.copy()` method was needed was because, any operation we carry out using the `fillna()` method amongst other methods is carried out on a view of the main DataFrame and affects the original values of the main df, but once we use .copy(), we make a copy of the df and carry out our operations on this copy.

**Explanation for negative effects of filling with 0 for averages**

If we fill missing values with 0, we inadvertently skew the mean to the left because our zero value pushes the overall average lower.
