### Exercise 2: Loops without shortcuts

````python
""" 
    The aim of this exercise is to perform 'sum()', 'min()' and 'max()' operations without the built in functions
    using only a basic 'for' loop and logic.
"""

los = [3, 5, 12, 8, 4, 15, 2, 6]

for i in range(len(los)):
    if i == 0:
        _sum = los[i]
        _min = los[i]
        _max = los[i]
    else:
      _sum += los[i]
      if los[i] < _min:
        _min = los[i]
      elif los[i] > _max:
        _max = los[i]

print(f"Sum: {_sum}")
print(f"Min: {_min}")
print(f"Max: {_max}")
````

### Exercise 3: Statistics by Hand

Write three functions using no libraries:


*   mean(values)
*   median(values)
*   sample_sd(values), dividing by n−1

Test on

```
dataset = [4, 7, 7, 9, 10, 12, 95]
```

You should get a mean of about **20.57**, a median of **9** and an SD of about **32.9**. If you don’t, find your bug before moving on.

````python
dataset = [4, 7, 7, 9, 10, 12, 95]

""" The function to determine the mean of a dataset goes here, 
    it just takes the sum of the stored values and divides it by the length of the list
"""
def mean(ds: list) -> float:
    total = 0
    if not ds:
        raise ValueError('Cannot take the mean of an empty list')
    for value in ds:
        total += value
    return total / len(ds)


mean(dataset)
````

````python
""" The function to determine the median value of a dataset goes here, 
    it operates in two ways, if length of the dataset is even, it sums the two middle values 
    and returns their average as the result, while if the length is odd, it just finds the index 
    of the middle value divides that by two and outputs the int value as the result
"""

def median(ds: list) -> float | int:
    if not ds:
        raise ValueError("Cannot take the median of an empty list")

    ds = sorted(ds) # Sorted return a copy so the original stays intact
    mid = len(ds) // 2 # To find the integer of the mid value I used floor division

    if len(ds) % 2 == 0:
        value = (ds[mid] + ds[mid - 1]) / 2
    else:
        value = ds[mid]
    return value

print(median(dataset))
print(median([1, 2, 3, 4]))
print(median([9, 1, 5]))
````

````python
""" The function to determine the sample standard deviation of a dataset goes here, 
    we first find the mean of the dataset and create a counter to store the squared difference,
    then we execute a loop to find (x - bar(x)) for each value in the dataset, square that value
    and store it in a diff variable, which is then concatenated to the sq_diff list, the sum taken
    and divided by n (which is the length of the dataset) - 1 to give us the deviation/variance
    whose sqrt() is taken to get our standard deviation
"""

def sample_sd(ds: list) -> float:
    if len(ds) < 2:
        raise ValueError("Cannot compute standard deviation for list of less than 2 values")
    
    mean_value = mean(ds)
    total = 0
    for data_point in ds:
        diff = data_point - mean_value
        total += (diff**2)
    variance = total / (len(ds) - 1)
    sd = variance ** 0.5
    return sd

print(sample_sd(dataset))
print(sample_sd([7, 9, 10]))
````

### Exercise 4: Group by Hand

Using loops and a dictionary, compute the mean length of stay and readmission rate for each ward. 
Expected: **A ≈ 4.67 days and 0.67**, **B = 8 days and 0.33**, **C = 8.5 days and 0.5**. This is what *groupby* does in ***Pandas***. If you write it by hand now, *groupby* will make sense tomorrow.

````python
"""
    The dataset we would be working on goes here:
"""

records = [
    {"ward": "A", "los": 3,  "readmitted": 0},
    {"ward": "A", "los": 5,  "readmitted": 1},
    {"ward": "B", "los": 12, "readmitted": 1},
    {"ward": "B", "los": 8,  "readmitted": 0},
    {"ward": "B", "los": 4,  "readmitted": 0},
    {"ward": "C", "los": 15, "readmitted": 1},
    {"ward": "C", "los": 2,  "readmitted": 0},
    {"ward": "A", "los": 6,  "readmitted": 1},
]
````

````python
def group_by(records: list) -> list:
    records_dict = {}
    overview = []
    for record in records:
        ward = record['ward']
        los = record['los']
        readmitted = record['readmitted']

        if ward not in records_dict:
            records_dict[ward] = {'los': [], 'readmitted': []}
        records_dict[ward]['los'].append(los)
        records_dict[ward]['readmitted'].append(readmitted)

    for ward, data in records_dict.items():
        mean_los = mean(data['los'])
        readmission_rate = mean(data['readmitted'])
        overview.append([ward, f'{mean_los:.2f} days', f'{readmission_rate:.2f}'])
        
    return overview

group_by(records=records)
````

````python
"""
    Just because I finished early, I am writing the function to detect Interquartile Range outliers with no libraries. 
"""
scores = [9, 8, 7, 7, 45, 12, 10, 9, 11, 6, 48]


def iqr_outliers(values: list):
    values = sorted(values)
    q1, q3 = values[int(len(values) * .25)], values[int(len(values) * .75)]
    for value in values:
        if (value < q1 - 1.5 * (q3 - q1)) or (value > q3 + 1.5 * (q3 - q1)):
            return f"{value} is an outlier"

iqr_outliers(scores)
````
