### Exercise 2: Loops without shortcuts

The aim of this exercise is to perform 'sum()', 'min()' and 'max()' operations without the built in functions,
using only a basic **if** and **for** loop alongside programming logic.

````python
los = [3, 5, 12, 8, 4, 15, 2, 6]
````

````python
_sum = 0
_min = 0
_max = 0
for i in range(len(los)):
    if i == 0:
        _sum = los[i]
        _min = los[i]
        _max = los[i]
    else:
      _sum += los[i]
      if los[i] < _min:
        _min = los[i]
      elif los[i] > _max:
        _max = los[i]

print(f"Sum: {_sum}")
print(f"Min: {_min}")
print(f"Max: {_max}")
````

````python
# MEAN FUNCTION

def mean(ds: list) -> float:
    """ 
        Return the arithmetic mean of a list of numbers.
         
        raise `ValueError` if the list is empty and `TypeError` if any
        element is not an int or float.
    """
    if not ds:
        raise ValueError('Cannot take the mean of an empty list')
    
    total = 0 
    for value in ds:
        if any(not isinstance(item, (int, float)) for item in ds):
            raise TypeError('Cannot add values of none numerical data type')
        total += value
    return total / len(ds)
````

````python
# MEDIAN FUNCTION

def median(ds: list) -> float | int:
    """ Returns the middle value in a list of numbers.
    
        It creates a sorted copy of the original list from which,
        it checks the length of the dataset.
        if `len(ds)` is even it returns the average of the two middle values.
        
        raises a `ValueError` for an empty list and a `TypeError` for non-numerical datatypes. 
    """

    if not ds:
        raise ValueError("Cannot take the median of an empty list")
    if any(not isinstance(item, (int, float)) for item in ds):
        raise TypeError('Cannot get median of non-numerical data type')
    
    ds = sorted(ds) # sorted() returns a copy so the original list stays intact
    mid = len(ds) // 2

    if len(ds) % 2 == 0:
        value = (ds[mid] + ds[mid - 1]) / 2
    else:
        value = ds[mid]
    return value
````

````python
# SAMPLE STANDARD DEVIATION FUNCTION

def sample_sd(ds: list) -> float:
    """ 
        Returns the standard deviation for a sample of a population.
        
        Divides the sum of the squared difference by (n - 1), it uses
        the mean function previously created to get the mean_value.
        
        raises a `ValueError` for a list with less than 2 values,
        and `TypeError` for str in list.
    """

    if len(ds) < 2:
        raise ValueError("Cannot compute standard deviation for list of less than 2 values")
    
    mean_value = mean(ds)
    total = 0
    for data_point in ds:
        diff = data_point - mean_value
        total += (diff**2)
    variance = total / (len(ds) - 1)
    sd = variance ** 0.5
    return sd
````

### Exercise 4: `Group` by hand

Using loops and a dictionary, compute the mean length of stay and readmission rate for each ward. 
Expected: **A ≈ 4.67 days and 0.67**, **B = 8 days and 0.33**, **C = 8.5 days and 0.5**. This is what *groupby* does in ***Pandas***. If you write it by hand now, *groupby* will make sense tomorrow.

````python
def summarise_by_ward(records: list[dict]) -> dict:
    """ Return the mean length of stay and readmission rate in a dictionary.

        The mean LOS and readmission rate is matched to each ward using
        the ward name as the dict key. This function takes a `list` of `dict` 
        as the input and outputs a `dict` of the mean length of stay 
        and readmission rate for each ward.
        
        An empty list returns {} without an error, while a record missing 
        the `ward` key raises a raw `KeyError`.   
    """
    
    records_dict = {}
    overview = {}
    for record in records:
        ward = record['ward']
        los = record['los']
        readmitted = record['readmitted']
        if not ward:
            raise KeyError('The ward key is missing from the dictionary')

        if ward not in records_dict:
            records_dict[ward] = {'los': [], 'readmitted': []}
        records_dict[ward]['los'].append(los)
        records_dict[ward]['readmitted'].append(readmitted)

    for ward, data in records_dict.items():
        mean_los = mean(data['los'])
        readmission_rate = mean(data['readmitted'])
        overview[ward] = {'mean_los': mean_los, 'readmission_rate': readmission_rate}
        
    return overview
````

### *The IQR Outlier Function*

````python

def iqr_outliers(values: list | tuple | set) -> list[int]:
    """ Return the outliers in values, sorted ascending.
     
        Uses the 1.5 x IQR rule. Quartiles are taken by index position on 
        the sorted list, with no interpolation, So results may differ from 
        Pandas/Numpy which interpolate by default.
    """
    
    if len(values) < 2:
        raise ValueError('Cannot calculate IQR outliers for empty list and list with only one value')
    if type(values) is dict:
        raise TypeError('Cannot perform IQR outlier calculation on dictionaries')
    for value in values:
        if not isinstance(value, (int, float)):
            raise TypeError('Cannot sort non-numerical data type')
    
    values = sorted(values)
    # The Quartiles are gotten by taking the index position on the sorted list (No interpolation)
    # this may differ from Pandas/Numpy which interpolate by default.
    q1, q3 = values[int(len(values) * .25)], values[int(len(values) * .75)]
    iqr = q3 - q1
    upper, lower = (q3 + 1.5 * iqr), (q1 - 1.5 * iqr)
    outliers = []
    for value in values:
        if value < lower or value > upper:
            outliers.append(value)
    
    return outliers
````
